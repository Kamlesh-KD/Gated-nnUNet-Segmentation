```python
import os
import gzip
import shutil
from dummy_modalities import create_dummy_modalities

# Set up nnU-Net environment variables
os.environ['nnUNet_raw'] = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw"
os.environ['nnUNet_preprocessed'] = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_preprocessed"
os.environ['nnUNet_results'] = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_results"

temp_folder = "/root/breast_seg/temp_folder"
input_image = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/Dataset001_BreastTumor/imagesTs/BreastMRI048_0000.nii"

if input_image.endswith('.nii'):
    output_path = os.path.join(temp_folder, os.path.basename(input_image).replace('.nii', '.nii.gz'))
    os.makedirs(temp_folder, exist_ok=True)
    with open(input_image, "rb") as f_in:
        with gzip.open(output_path, "wb") as f_out:
            shutil.copyfileobj(f_in, f_out)
    input_image = output_path

create_dummy_modalities(input_image, 4, temp_folder)
```

```python
input_dir = os.path.join(temp_folder, os.path.basename(input_image).replace('.nii.gz', ''))
output_dir = os.path.join("/root/breast_seg/single_inference/", os.path.basename(input_image).replace('.nii.gz', ''))
dataset_id = "Dataset001_BreastTumor"
config = "3d_fullres"
fold = 4
 
cmd = f"""
nnUNetv2_predict \
-i {input_dir} \
-o {output_dir} \
-d {dataset_id} \
-c {config} \
-f {fold} \
-chk checkpoint_best.pth
"""
print("Running:\n", cmd)
os.system(cmd)
```
