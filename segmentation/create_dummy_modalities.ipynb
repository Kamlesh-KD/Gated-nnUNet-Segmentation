```python
import os
import nibabel as nib
import numpy as np
from pathlib import Path
import shutil

def create_dummy_modality(input_file, output_file):
    """
    Create a dummy modality by copying the input file.
    In a more sophisticated approach, you could:
    - Use zeros
    - Use noise
    - Apply simple transformations
    - Copy the original modality
    """
    # We'll copy the original modality as a simple approach
    # This preserves the basic structure and intensity distribution
    shutil.copy2(input_file, output_file)
    print(f"Created dummy modality: {output_file}")

def process_single_modality_directory(input_dir, output_dir):
    """
    Process a directory containing single-modality images and create 4-modality versions.
    """
    input_path = Path(input_dir)
    output_path = Path(output_dir)
    
    # Create output directory if it doesn't exist
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Find all single-modality files (should be _0000)
    single_modality_files = list(input_path.glob("*_0000.nii*"))
    
    print(f"Found {len(single_modality_files)} single-modality files")
    
    for file_path in single_modality_files:
        # Extract base name (case identifier)
        base_name = file_path.stem.replace('_0000', '')
        
        # Create output files for all 4 modalities
        for modality in range(4):
            output_file = output_path / f"{base_name}_{modality:04d}.nii.gz"
            
            if modality == 0:
                # Copy the original modality
                if file_path.suffix == '.gz':
                    shutil.copy2(file_path, output_file)
                else:
                    # Convert .nii to .nii.gz if needed
                    img = nib.load(file_path)
                    nib.save(img, output_file)
                print(f"Copied original modality: {output_file}")
            else:
                # Create dummy modality
                create_dummy_modality(file_path, output_file)

```

```python
single_modality_dir = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/Dataset001_BreastTumor/imagesTs"
four_modality_dir = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/Dataset001_BreastTumor/imagesTs_4mod"

print("Processing single-modality images...")
print(f"Input directory: {single_modality_dir}")
print(f"Output directory: {four_modality_dir}")

process_single_modality_directory(single_modality_dir, four_modality_dir)

print("\nProcessing complete!")
print(f"4-modality images saved to: {four_modality_dir}")
print("You can now use this directory for inference.")
```

```python
import os
import nibabel as nib
import numpy as np
from pathlib import Path
import shutil

def create_dummy_modalities(input_file_path, number_of_modalities, output_path="/root/breast_seg/dummy_modalities"):
    """
    Create multiple dummy modalities by copying a single image file.
    Saves to {output_path}/{filename_without_extension}/{filename_without_extension}_{modality_number}.nii.gz
    
    Args:
        input_file_path: Path to the input image file (e.g., .nii.gz or .nii)
        number_of_modalities: Number of dummy modalities to create
        output_path: Base output directory path
    
    Returns:
        List of paths to the output files
    """
    input_path = Path(input_file_path)
    
    # Extract file name without extension
    file_name_without_extension = input_path.stem
    if input_path.suffix == '.gz':
        # Handle .nii.gz case
        file_name_without_extension = Path(input_path.stem).stem
    
    # Create output directory structure
    output_dir = Path(output_path) / file_name_without_extension
    output_dir.mkdir(parents=True, exist_ok=True)
    
    output_files = []
    
    # Create dummy modalities
    for modality_num in range(number_of_modalities):
        output_file = output_dir / f"{file_name_without_extension}_{modality_num:04d}.nii.gz"
        
        # Copy the file
        shutil.copy2(input_path, output_file)
        print(f"Created dummy modality {modality_num}: {output_file}")
        output_files.append(output_file)
    
    return output_files
```

```python
# Example: Process a single image file and create 4 dummy modalities
input_image = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/test_labels/BreastMRI287.nii.gz"
num_modalities = 4

output_files = create_dummy_modalities(input_image, num_modalities)

print("\nProcessing complete!")
print(f"Dummy modalities saved to:")
for f in output_files:
    print(f"  - {f}")
```
