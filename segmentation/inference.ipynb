```python
import shutil
from pathlib import Path
 
# Directory containing files to rename
val_d_dir = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/Dataset001_BreastTumor/val_d"
 
# Get all files in the directory
files = list(Path(val_d_dir).glob("*.nii.gz"))
 
print(f"Found {len(files)} files to rename")
 
for file_path in files:
    # Current format: Breast_MRI_XXX_YY.nii.gz
    # Target format: BreastMRIXXX_YYYY.nii.gz
 
    filename = file_path.name
 
    # Extract components
    parts = filename.replace('.nii.gz', '').split('_')
 
    # parts should be: ['Breast', 'MRI', 'XXX', 'YY']
    if len(parts) == 4:
        case_id = parts[2]  # XXX
        modality = parts[3]  # YY
 
        # Convert to target format
        new_filename = f"BreastMRI{case_id}_{int(modality):04d}.nii.gz"
 
        # Rename the file
        old_path = file_path
        new_path = file_path.parent / new_filename
 
        shutil.move(old_path, new_path)
        print(f"Renamed: {filename} -> {new_filename}")
    else:
        print(f"Skipping {filename} - unexpected format")
 
print("\nRenaming complete!")

```

```python
import shutil
from pathlib import Path

# Source directory containing _0000.nii.gz files
source_dir = "/root/breast_seg/nnUNet_Frame/dataset/nnUNet_raw/Dataset001_BreastTumor/val_d"

# Target directory
target_dir = "/root/RL/dd"

# Create target directory if it doesn't exist
Path(target_dir).mkdir(parents=True, exist_ok=True)

# Find all files ending with _0000.nii.gz
files_to_copy = list(Path(source_dir).glob("*_0000.nii.gz"))

print(f"Found {len(files_to_copy)} files ending with _0000.nii.gz")

# Copy the files
for file_path in files_to_copy:
    target_path = Path(target_dir) / file_path.name
    shutil.copy2(file_path, target_path)
    print(f"Copied: {file_path.name} -> {target_dir}")

print(f"\nCopying complete! {len(files_to_copy)} files copied to {target_dir}")
```
