check mask for each image

- if all values 0, only background
- if has some other values, has lesion

```python
import SimpleITK as sitk
import numpy as np
import glob
fs = sorted(glob.glob('/workspace/Dataset002_dartmouth_unilateral_mixed/labelsTr/*.nii.gz'))
vals, pos = set(), 0
for f in fs:
    a = sitk.GetArrayFromImage(sitk.ReadImage(f)); u = np.unique(a); vals |= set(u.tolist()); pos += (u > 0).any()
print(len(fs), 'masks | with lesion:', pos, '| without:', len(fs) - pos, '| label values:', vals)
```
